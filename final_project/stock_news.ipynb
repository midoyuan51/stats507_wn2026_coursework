```python
import pandas as pd
import os
import numpy as np
import matplotlib.pyplot as plt

INPUT_FILE    = "nasdaq_exteral_data.csv"
OUTPUT_FILE   = "all_stocks_data.csv"
CHUNK_SIZE    = 200_000

# Updated ticker list
TARGET_STOCKS = [
    # Tech (large cap)
    'AAPL', 'AMZN', 'MSFT', 'TSLA', 'GOOGL', 'NVDA', 'NFLX',
    # Finance
    'JPM', 'BAC', 'GS', 'MS', 'WFC',
    # Healthcare/Pharma
    'PFE', 'JNJ', 'ABT', 'MRK', 'ABBV', 'GILD',
    # Energy
    'XOM', 'CVX', 'COP', 'SLB',
    # Retail/Consumer
    'WMT', 'TGT', 'NKE', 'COST', 'MCD',
    # EV/Clean Energy
    'NIO', 'F', 'GM', 'ENPH',
    # Semiconductor
    'AMD', 'INTC', 'QCOM', 'AVGO',
    # Aerospace/Defense
    'BA', 'LMT', 'RTX',
    # Real Estate
    'AMT', 'PLD', 'PSA', 'DLR',
]

first_chunk = True

for i, chunk in enumerate(pd.read_csv(INPUT_FILE, chunksize=CHUNK_SIZE,
                                       usecols=['Date', 'Article_title', 'Lsa_summary', 'Luhn_summary', 'Textrank_summary', 'Stock_symbol'])):
    filtered = chunk[chunk['Stock_symbol'].isin(TARGET_STOCKS)]
    if len(filtered) > 0:
        mode = 'w' if first_chunk else 'a'
        filtered.to_csv(OUTPUT_FILE, index=False, mode=mode, header=first_chunk)
        first_chunk = False

    if i % 5 == 0:  # print every 500k rows instead of every 1M
        print(f"Processed {i * CHUNK_SIZE:,} rows...")

print("Done!")
print(pd.read_csv(OUTPUT_FILE)['Stock_symbol'].value_counts())
```
